```python
import requests
import json
from bs4 import BeautifulSoup
import pandas as pd

# Charger les catégories depuis le fichier JSON
with open("config/categories.json", "r") as f:
    categories = json.load(f)

# Afficher les catégories
categories

# Choisir une catégorie (ex: téléphones & tablettes)
url = categories["telephones_tablettes"]

# Envoyer une requête GET à la page
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
response = requests.get(url, headers=headers)

# Vérifier que la requête a fonctionné
print(f"Statut : {response.status_code}")

# Afficher les 1000 premiers caractères du HTML
print(response.text[:1000])


# Parser le HTML avec BeautifulSoup
soup = BeautifulSoup(response.text, "lxml")

# Trouver tous les blocs de produits (généralement des balises <article> ou <div>)
products = soup.find_all("article")

print(f"{len(products)} produits trouvés sur la page.")


# Extraire les informations de chaque produit
product_list = []

extracted_data = []

for product in products:
    # Nom du produit
    name_tag = product.find("h3")
    name = name_tag.text.strip() if name_tag else "N/A"
    
    # Lien produit
    link_tag = product.find("a", href=True)
    link = "https://www.jumia.ci" + link_tag["href"] if link_tag else "N/A"
    
    # Prix actuel
    price_tag = product.find("div", class_="prc")
    price = price_tag.text.strip() if price_tag else "N/A"
    
    # Ancien prix
    old_price_tag = product.find("div", class_="old")
    old_price = old_price_tag.text.strip() if old_price_tag else "N/A"
    
    # Avis / Évaluation
    rating_tag = product.find("div", class_="stars _s")
    rating = rating_tag.text.strip() if rating_tag else "N/A"

    # On stocke tout dans un dictionnaire
    extracted_data.append({
        "name": name,
        "price": price,
        "old_price": old_price,
        "link": link,
        "rating": rating
    })


import pandas as pd

df = pd.DataFrame(extracted_data)
df.head()

print(products[1].prettify())

# afficher les 100 premiers produits
df.to_csv("data/smartphones.csv", index=False)

df.head(50)

```
